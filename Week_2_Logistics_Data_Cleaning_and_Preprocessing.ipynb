## **3. Missing-Value Handling**

**3.1 Standardize Missing Values**

Hidden missing-value placeholders are converted into standard NaN values for consistent processing.

```python
# Define common missing-value placeholders

missing_placeholders = [
    "", " ", "NA", "N/A", "NULL",
    "null", "NaN", "nan", "None"
]

# Convert hidden missing values to NaN

for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].replace(missing_placeholders, np.nan)

    df[column] = df[column].apply(
        lambda x: np.nan
        if isinstance(x, str) and x.strip() == ""
        else x
    )

print("Hidden missing values standardized successfully.")
```

### **5. Data-Type Conversion and Cleaning**

**5.1 Convert Numeric Columns**


Numeric fields stored as text are converted to numeric data types for accurate calculations and analysis.

```python
# Convert numeric columns

numeric_columns = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "multiple_deliveries"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Clean and convert Time_taken(min)
df["Time_taken(min)"] = (
    df["Time_taken(min)"]
    .astype(str)
    .str.extract(r"(\d+)", expand=False)
)

df["Time_taken(min)"] = pd.to_numeric(
    df["Time_taken(min)"],
    errors="coerce"
)

print("Numeric columns converted successfully.")
```

**5.2 Convert Order Date**

The order date is converted to a datetime format to support date-based logistics analysis.

```python
# Convert order date to datetime

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    dayfirst=True,
    errors="coerce"
)

print("Order date converted successfully.")
```

**5.3 Convert Order Times**


Order and pickup times are converted into time-compatible formats for delivery-time analysis.

```python
# Convert order and pickup times to datetime format

df["Time_Orderd"] = pd.to_datetime(
    df["Time_Orderd"],
    format="%H:%M:%S",
    errors="coerce"
)

df["Time_Order_picked"] = pd.to_datetime(
    df["Time_Order_picked"],
    format="%H:%M:%S",
    errors="coerce"
)

print("Order and pickup times converted successfully.")
```

**5.4 Validate Data-Type Conversion**


Converted columns are validated to confirm that the required data types were applied successfully.

```python
# Validate converted data types

converted_columns = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "multiple_deliveries",
    "Time_taken(min)",
    "Order_Date",
    "Time_Orderd",
    "Time_Order_picked"
]

print(df[converted_columns].dtypes)
```

 **5.5 Inspect Converted Order Times**

**5.6 Treat Invalid Ratings**

Delivery ratings are validated against the expected 1–5 scale. Ratings above 5 are treated as invalid and converted to NaN so that they can be handled consistently with other missing values.

```python
# Identify invalid delivery ratings

invalid_ratings = (
    df["Delivery_person_Ratings"] > 5
).sum()

print("Invalid ratings above 5:", invalid_ratings)

# Convert invalid ratings to NaN

df.loc[
    df["Delivery_person_Ratings"] > 5,
    "Delivery_person_Ratings"
] = np.nan

print(
    "Missing ratings after invalid-value treatment:",
    df["Delivery_person_Ratings"].isnull().sum()
)
```

**5.7 Handle Missing Values After Data-Type Conversion**


Missing numerical values are treated using median imputation to reduce the influence of extreme observations. Missing categorical values are treated using mode imputation. Missing order times are handled using the most frequent valid time.

```python
# Handle missing numerical values using median imputation

numerical_missing_columns = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "multiple_deliveries"
]

for column in numerical_missing_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(
            df[column].median()
        )


# Handle missing categorical values using mode imputation

categorical_missing_columns = [
    "Road_traffic_density",
    "Festival",
    "City"
]

for column in categorical_missing_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(
            df[column].mode()[0]
        )


# Handle missing order times using the most frequent valid time

if df["Time_Orderd"].isnull().sum() > 0:
    df["Time_Orderd"] = df["Time_Orderd"].fillna(
        df["Time_Orderd"].mode()[0]
    )

print("Missing-value treatment completed.")
```

### **6. Outlier Detection and Treatment**

**6.1 Identify Numerical Columns**


The Interquartile Range (IQR) method is used to identify potential outliers in key numerical variables. These observations are reviewed before treatment because extreme values may represent genuine logistics conditions.

```python
# Identify potential outliers using the IQR method

outlier_columns = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "multiple_deliveries",
    "Time_taken(min)"
]

outlier_summary = []

for column in outlier_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Potential Outliers": outlier_count
    })

outlier_summary_df = pd.DataFrame(outlier_summary)

outlier_summary_df
```

**6.2 Review Potential Outliers**

IQR screening identified potential extreme observations. These values are reviewed against the expected business range before treatment. Values that are operationally valid are retained rather than removed automatically.

```python
# Validate key variables against their expected business ranges

range_validation = pd.DataFrame({
    "Variable": [
        "Delivery_person_Age",
        "Delivery_person_Ratings",
        "multiple_deliveries",
        "Time_taken(min)"
    ],
    "Minimum": [
        df["Delivery_person_Age"].min(),
        df["Delivery_person_Ratings"].min(),
        df["multiple_deliveries"].min(),
        df["Time_taken(min)"].min()
    ],
    "Maximum": [
        df["Delivery_person_Age"].max(),
        df["Delivery_person_Ratings"].max(),
        df["multiple_deliveries"].max(),
        df["Time_taken(min)"].max()
    ]
})

range_validation
```

**6.3 — Outlier Treatment Decision**


Based on range validation, the identified extreme observations fall within plausible operational ranges. Therefore, no records are removed solely based on IQR screening. Retaining these observations preserves genuine variations in delivery operations and avoids losing useful information for subsequent logistics analysis.

```python
# Confirm that key numerical variables contain no invalid range values

invalid_range_counts = {
    "Delivery_person_Age": (
        (df["Delivery_person_Age"] < 15) |
        (df["Delivery_person_Age"] > 50)
    ).sum(),

    "Delivery_person_Ratings": (
        (df["Delivery_person_Ratings"] < 1) |
        (df["Delivery_person_Ratings"] > 5)
    ).sum(),

    "multiple_deliveries": (
        (df["multiple_deliveries"] < 0) |
        (df["multiple_deliveries"] > 3)
    ).sum(),

    "Time_taken(min)": (
        (df["Time_taken(min)"] < 10) |
        (df["Time_taken(min)"] > 54)
    ).sum()
}

invalid_range_counts
```

**6.4 Validate Outlier Treatment**

The outlier treatment is validated to confirm that delivery ratings now fall within the expected 1–5 range and that no missing ratings remain after imputation. Legitimate extreme values in other numerical variables are retained.


```python
# Validate rating treatment

print("Ratings above 5:", (df["Delivery_person_Ratings"] > 5).sum())
print("Ratings below 1:", (df["Delivery_person_Ratings"] < 1).sum())
print("Missing ratings:", df["Delivery_person_Ratings"].isnull().sum())
```
